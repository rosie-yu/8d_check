```python
%load_ext autoreload
%autoreload 2
import pandas as pd
from eightd_evaluator import evaluate, classify_score
```

### 讀資料

```python
# 評分標準
score_guide = pd.read_excel('data/8d_data.xlsx', sheet_name='評核標準')
score_guide.drop(columns='項目目的', inplace=True)
score_guide
```

```python
# 資料
report1 = pd.read_excel('data/8d_data.xlsx', sheet_name='8D報告內容')
report1
```

### 評分

```python
# 欄位缺少
score_items = ['test']
print('欲評分欄位:', score_items)
evaluate(data=report1, guide=score_guide, check_item=score_items)
```

```python
# 欄位齊全
score_items = list(score_guide['項目'])
print('欲評分欄位:', score_items)
report1_scored = evaluate(data=report1, guide=score_guide, check_item=score_items)
report1_scored
```

```python
# 分類評分(差劣/普通/不錯/傑出)
report1_scored['score_level'] = report1_scored['score'].apply(classify_score)
display(report1_scored)
```

### 存檔

```python
report1_scored.to_excel('data/scored_report.xlsx')
```
